```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import TimeSeriesSplit

step = 4
lookback = 12
kfold = 5

# =========================
# DATA
# =========================
Xx = data.drop(columns=['Target'])
Yy = data['Target']

# =========================
# 1. CREATE SEQUENCE + META (FIXED TYPE)
# =========================
def create_sequence_full(Xx, Yy, lookback, step):
    X_all, Y_all, meta = [], [], []

    provinces = Xx.index.get_level_values('Province').unique()

    for prov in provinces:
        X_p = Xx.loc[prov]
        Y_p = Yy.loc[prov]

        for i in range(len(X_p) - lookback - step + 1):
            X_seq = X_p.iloc[i:i+lookback].values
            Y_seq = Y_p.iloc[i+lookback:i+lookback+step].values

            year = int(X_p.iloc[i+lookback-1].name[0])
            month = int(X_p.iloc[i+lookback-1].name[1])

            X_all.append(X_seq)
            Y_all.append(Y_seq)
            meta.append((prov, year, month))

    return np.array(X_all), np.array(Y_all), np.array(meta, dtype=object)

# generate
X_seq, Y_seq, meta = create_sequence_full(Xx, Yy, lookback, step)

# =========================
# 2. SORT GLOBAL TIME (FIXED)
# =========================
years = meta[:,1].astype(int)
months = meta[:,2].astype(int)

order = np.lexsort((months, years))

X_seq = X_seq[order]
Y_seq = Y_seq[order]
meta = meta[order]

# =========================
# 3. FORWARD CHAINING SPLIT
# =========================
tscv = TimeSeriesSplit(n_splits=kfold)

X_train_fold, X_test_fold = {}, {}
Y_train_fold, Y_test_fold = {}, {}

results = []

for fold, (train_idx, test_idx) in enumerate(tscv.split(X_seq)):

    X_train, X_test = X_seq[train_idx], X_seq[test_idx]
    y_train, y_test = Y_seq[train_idx], Y_seq[test_idx]

    meta_train = meta[train_idx]
    meta_test = meta[test_idx]

    # =========================
    # SCALING (NO LEAKAGE)
    # =========================
    n_train, lb, nf = X_train.shape
    n_test = X_test.shape[0]

    scaler = StandardScaler()

    X_train_2d = X_train.reshape(-1, nf)
    X_test_2d  = X_test.reshape(-1, nf)

    before_min = X_train_2d.min()
    before_max = X_train_2d.max()

    X_train_scaled = scaler.fit_transform(X_train_2d)
    X_test_scaled  = scaler.transform(X_test_2d)

    after_min = X_train_scaled.min()
    after_max = X_train_scaled.max()

    # reshape back
    X_train_scaled = X_train_scaled.reshape(n_train, lb, nf)
    X_test_scaled  = X_test_scaled.reshape(n_test, lb, nf)

    # save fold
    X_train_fold[fold] = X_train_scaled
    X_test_fold[fold] = X_test_scaled
    Y_train_fold[fold] = y_train
    Y_test_fold[fold] = y_test

    # =========================
    # DIAGNOSTICS
    # =========================
    train_years = meta_train[:,1].astype(int)
    test_years = meta_test[:,1].astype(int)

    train_prov = np.unique(meta_train[:,0])
    test_prov = np.unique(meta_test[:,0])

    print(f"\n===== FOLD {fold} =====")

    print(f"Lookback window : {lookback} months (t-12 → t-1)")
    print(f"Horizon target  : {step} months (t → t+{step-1})")

    print("Train years:", np.unique(train_years))
    print("Test years :", np.unique(test_years))
    print("Train provinces:", len(train_prov))
    print("Test provinces :", len(test_prov))
    print("Scaling before:", (before_min, before_max))
    print("Scaling after :", (after_min, after_max))

    results.append({
        "fold": fold,
        "train_min_year": train_years.min(),
        "train_max_year": train_years.max(),
        "test_min_year": test_years.min(),
        "test_max_year": test_years.max(),
        "train_size": len(train_idx),
        "test_size": len(test_idx)
    })

# =========================
# 4. PLOT FORWARD CHAINING
# =========================
# df_plot = pd.DataFrame(results)

# plt.figure()

# for i, row in df_plot.iterrows():
#     plt.barh(i, row['train_size'], left=0)
#     plt.barh(i, row['test_size'], left=row['train_size'])

# plt.yticks(range(kfold), [f"Fold {i}" for i in range(kfold)])
# plt.xlabel("Number of Samples")
# plt.ylabel("Fold")
# plt.title("Forward Chaining Split (Train vs Test)")

# plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    accuracy_score, f1_score, recall_score, precision_score,
    classification_report, confusion_matrix,
    average_precision_score, matthews_corrcoef,
    precision_recall_curve, auc
)

def eval_fold(Y_true, Y_pred, Y_prob):
    
    # =========================
    # INIT METRICS
    # =========================
    eval_metric = {
        'accuracies': [], 
        'f1_macro': [],
        'f1_micro': [],
        'recalls': [], 
        'precisions': [],
        'pr_aucs': [],
        'mccs': []
    }
    
    y_true_all = []
    y_pred_all = []
    y_prob_all = []  # 🔥 tambahan
    
    # =========================
    # LOOP PER FOLD
    # =========================
    for i in range(len(Y_true)):
        
        y_true = Y_true[i].flatten()
        y_pred = Y_pred[i].flatten()
        y_prob = Y_prob[i].flatten()
    
        y_true_all.extend(y_true)
        y_pred_all.extend(y_pred)
        y_prob_all.extend(y_prob)  # 🔥 tambahan
    
        # --- METRICS ---
        acc = accuracy_score(y_true, y_pred)
        f1_macro = f1_score(y_true, y_pred, average='macro')
        f1_micro = f1_score(y_true, y_pred, average='micro')
        recall = recall_score(y_true, y_pred, average='macro')
        precision = precision_score(y_true, y_pred, average='macro')
        
        pr_auc = average_precision_score(y_true, y_prob)
        mcc = matthews_corrcoef(y_true, y_pred)
    
        # --- SAVE ---
        eval_metric['accuracies'].append(acc)
        eval_metric['f1_macro'].append(f1_macro)
        eval_metric['f1_micro'].append(f1_micro)
        eval_metric['recalls'].append(recall)
        eval_metric['precisions'].append(precision)
        eval_metric['pr_aucs'].append(pr_auc)
        eval_metric['mccs'].append(mcc)
    
        # --- PRINT PER FOLD ---
        print('=' * 40)
        print(f"Evaluasi untuk Fold ke-{i+1}:")
        print(classification_report(y_true, y_pred))
        print(f"PR AUC   : {pr_auc:.4f}")
        print(f"MCC      : {mcc:.4f}")
        print(f"F1 Micro : {f1_micro:.4f}")
        print('=' * 40)
    
    # =========================
    # MEAN & STD
    # =========================
    def mean_std(x):
        return np.mean(x), np.std(x, ddof=1)
    
    mean_acc, std_acc = mean_std(eval_metric['accuracies'])
    mean_f1_macro, std_f1_macro = mean_std(eval_metric['f1_macro'])
    mean_f1_micro, std_f1_micro = mean_std(eval_metric['f1_micro'])
    mean_recall, std_recall = mean_std(eval_metric['recalls'])
    mean_precision, std_precision = mean_std(eval_metric['precisions'])
    mean_pr_auc, std_pr_auc = mean_std(eval_metric['pr_aucs'])
    mean_mcc, std_mcc = mean_std(eval_metric['mccs'])
    
    # =========================
    # PRINT SUMMARY
    # =========================
    print()
    print('='*10, 'HASIL RATA - RATA KESELURUHAN','='*10)
    print("=" * 50)
    print(f" Accuracy   : {mean_acc:.4f} ± {std_acc:.4f}")
    print(f" F1 Macro   : {mean_f1_macro:.4f} ± {std_f1_macro:.4f}")
    print(f" F1 Micro   : {mean_f1_micro:.4f} ± {std_f1_micro:.4f}")
    print(f" Recall     : {mean_recall:.4f} ± {std_recall:.4f}")
    print(f" Precision  : {mean_precision:.4f} ± {std_precision:.4f}")
    print(f" PR AUC     : {mean_pr_auc:.4f} ± {std_pr_auc:.4f}")
    print(f" MCC        : {mean_mcc:.4f} ± {std_mcc:.4f}")
    print("=" * 50)
    
    # =========================
    # CONFUSION MATRIX GLOBAL
    # =========================
    cm = confusion_matrix(y_true_all, y_pred_all)
    
    plt.figure(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=['Not Late', 'Late'],
                yticklabels=['Not Late', 'Late'])
    plt.title(f'Confusion Matrix')
    plt.xlabel('Predicted Label')
    plt.ylabel('True Label')
    plt.show()
    
    # =========================
    # 🔥 PR CURVE GLOBAL (ALL FOLD)
    # =========================
    y_true_all = np.array(y_true_all)
    y_prob_all = np.array(y_prob_all)

    # precision, recall, _ = precision_recall_curve(y_true_all, y_prob_all)
    # pr_auc_global = auc(recall, precision)

    # plt.figure(figsize=(6,5))
    # plt.plot(recall, precision, label=f'PR-AUC = {pr_auc_global:.4f}')
    # plt.xlabel('Recall')
    # plt.ylabel('Precision')
    # plt.title(f'PR Curve - {nama_model}')
    # plt.legend()
    # plt.grid(True)
    # plt.show()
    
    # =========================
    # DATAFRAME PER FOLD
    # =========================
    df_CV_all_province = pd.DataFrame({
        f'Fold {i+1}': [
            eval_metric['accuracies'][i],
            eval_metric['f1_macro'][i],
            eval_metric['f1_micro'][i],
            eval_metric['recalls'][i],
            eval_metric['precisions'][i],
            eval_metric['pr_aucs'][i],
            eval_metric['mccs'][i]
        ] for i in range(len(Y_true))
    }, index=['Accuracy', 'F1 Macro', 'F1 Micro', 'Recall', 'Precision', 'PR AUC', 'MCC'])
    
    # =========================
    # DATAFRAME MEAN ± STD
    # =========================
    df_MEAN_all_province = pd.DataFrame({
        'Mean': [
            mean_acc, mean_f1_macro, mean_f1_micro,
            mean_recall, mean_precision,
            mean_pr_auc, mean_mcc
        ],
        'STD': [
            std_acc, std_f1_macro, std_f1_micro,
            std_recall, std_precision,
            std_pr_auc, std_mcc
        ]
    }, index=['Accuracy', 'F1 Macro', 'F1 Micro', 'Recall', 'Precision', 'PR AUC', 'MCC'])
    
    print("\nDataFrame per Fold:")
    print(df_CV_all_province)
    
    print("\nDataFrame Mean ± STD:")
    print(df_MEAN_all_province)

    return df_CV_all_province, df_MEAN_all_province, y_true_all, y_prob_all
```

```python
from tensorflow.keras.layers import MultiHeadAttention, LayerNormalization, GlobalAveragePooling1D
from tensorflow.keras.layers import Input
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Lambda, Add
```

```python
print('='*10, 'FORWARD CHAINING CROSS VALIDATION -INFORMER- ','='*10)

y_true = {}
y_preds = {}
y_probs = {}

# =========================
# APPROX PROBSPARSE ATTENTION
# =========================
from tensorflow.keras.layers import Lambda

def prob_sparse_attention(x, num_heads=4, key_dim=32, top_k=20):
    attn = MultiHeadAttention(
        num_heads=num_heads,
        key_dim=key_dim
    )(x, x)

    # =========================
    # SAFE KERAS OPS
    # =========================
    def sparse_fn(attn_tensor):
        scores = tf.reduce_mean(attn_tensor, axis=-1)

        k = tf.minimum(top_k, tf.shape(scores)[-1])

        values, indices = tf.math.top_k(scores, k=k)

        mask = tf.reduce_sum(
            tf.one_hot(indices, depth=tf.shape(scores)[-1]),
            axis=1
        )

        mask = tf.expand_dims(mask, axis=-1)

        return attn_tensor * mask

    attn_sparse = Lambda(sparse_fn)(attn)

    return attn_sparse


def create_informer_model(input_shape, output_dim):
    inputs = Input(shape=input_shape)

    # =========================
    # PROJECTION
    # =========================
    x = Dense(32)(inputs)

    # =========================
    # INFORMER ATTENTION BLOCK
    # =========================
    attn_output = prob_sparse_attention(x)

    x = LayerNormalization(epsilon=1e-6)(x + attn_output)

    # =========================
    # FEED FORWARD
    # =========================
    x_ff = Dense(64, activation='relu')(x)
    x_ff = Dropout(0.1)(x_ff)
    x_ff = Dense(32)(x_ff)

    x = LayerNormalization(epsilon=1e-6)(x + x_ff)

    # =========================
    # OPTIONAL SECOND BLOCK (biar lebih kuat)
    # =========================
    attn_output2 = prob_sparse_attention(x)
    x = LayerNormalization(epsilon=1e-6)(x + attn_output2)

    # =========================
    # POOLING
    # =========================
    x = GlobalAveragePooling1D()(x)

    # =========================
    # OUTPUT
    # =========================
    x = Dense(32, activation='relu')(x)
    x = Dropout(0.1)(x)
    outputs = Dense(output_dim, activation='sigmoid')(x)

    model = Model(inputs, outputs)

    model.compile(
        optimizer=Adam(learning_rate=0.001),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    return model


# =========================
# TRAINING LOOP (UNCHANGED)
# =========================
for i in range(5):
    print()
    print(f'================= FOLD = {i} ==============')
    print()

    X_train = X_train_fold[i]
    X_test = X_test_fold[i]
    y_train = Y_train_fold[i]
    y_test = Y_test_fold[i]

    # =========================
    # VALIDATION SPLIT
    # =========================
    split_idx = int(len(X_train) * 0.8)

    X_tr = X_train[:split_idx]
    X_val = X_train[split_idx:]
    
    y_tr = y_train[:split_idx]
    y_val = y_train[split_idx:]

    print("X_train.shape", X_tr.shape)
    print("X_val.shape", X_val.shape)
    print("X_test.shape", X_test.shape)

    # =========================
    # MODEL
    # =========================
    model = create_informer_model(X_train.shape[1:], y_train.shape[1])

    model.fit(
        X_tr, y_tr,
        epochs=epoch_global,
        batch_size=batch_size_global,
        validation_data=(X_val, y_val),
        callbacks=[early_stopping, Regularisasi],
        verbose=1
    )

    # =========================
    # PREDICTION
    # =========================
    y_pred_prob = model.predict(X_test).ravel()
    y_pred = (y_pred_prob >= threshold).astype(int)

    # =========================
    # SAVE
    # =========================
    y_true[i] = y_test
    y_preds[i] = y_pred
    y_probs[i] = y_pred_prob


# =========================
# EVALUATION
# =========================
df_cv_informer, df_meancv_informer, y_true_informer, y_prob_informer = eval_fold(y_true, y_preds, y_probs)
```

## 6) FEDFORMER

```python
from tensorflow.keras.layers import (
    Input, Dense, Dropout, LayerNormalization,
    GlobalAveragePooling1D, Lambda, Add
)
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam
import tensorflow as tf
```

```python
print('='*10, 'FORWARD CHAINING CROSS VALIDATION -FEDFORMER- ','='*10)

y_true = {}
y_preds = {}
y_probs = {}


from tensorflow.keras.layers import Lambda

# =========================
# DECOMPOSITION
# =========================
def series_decomposition(x):
    def decompose(tensor):
        trend = tf.reduce_mean(tensor, axis=1, keepdims=True)
        seasonal = tensor - trend
        return trend, seasonal

    trend, seasonal = Lambda(lambda t: decompose(t))(x)
    return trend, seasonal


# =========================
# FREQUENCY BLOCK (FFT)
# =========================
def frequency_block(x):
    def fft_fn(tensor):
        fft = tf.signal.fft(tf.cast(tensor, tf.complex64))
        fft_filtered = fft * tf.cast(0.5, tf.complex64)  # simple filter
        ifft = tf.signal.ifft(fft_filtered)
        # return tf.math.real(ifft)
        return tf.cast(tf.math.real(ifft), tf.float32)

    return Lambda(fft_fn)(x)


def create_fedformer_model(input_shape, output_dim):
    inputs = Input(shape=input_shape)

    # =========================
    # PROJECTION
    # =========================
    x = Dense(32)(inputs)

    # =========================
    # DECOMPOSITION
    # =========================
    trend, seasonal = series_decomposition(x)

    # =========================
    # FREQUENCY LEARNING
    # =========================
    seasonal_freq = frequency_block(seasonal)

    # =========================
    # COMBINE
    # =========================
    x = Add()([trend, seasonal_freq])

    # =========================
    # FEED FORWARD
    # =========================
    x_ff = Dense(64, activation='relu')(x)
    x_ff = Dropout(0.1)(x_ff)
    x_ff = Dense(32)(x_ff)

    x = LayerNormalization(epsilon=1e-6)(x + x_ff)

    # =========================
    # SECOND BLOCK (optional)
    # =========================
    trend2, seasonal2 = series_decomposition(x)
    seasonal_freq2 = frequency_block(seasonal2)
    x = Add()([trend2, seasonal_freq2])

    # =========================
    # POOLING
    # =========================
    x = GlobalAveragePooling1D()(x)

    # =========================
    # OUTPUT
    # =========================
    x = Dense(32, activation='relu')(x)
    x = Dropout(0.1)(x)
    outputs = Dense(output_dim, activation='sigmoid')(x)

    model = Model(inputs, outputs)

    model.compile(
        optimizer=Adam(learning_rate=0.001),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    return model


# =========================
# TRAINING LOOP (SAMA)
# =========================
for i in range(5):
    print()
    print(f'================= FOLD = {i} ==============')
    print()

    X_train = X_train_fold[i]
    X_test = X_test_fold[i]
    y_train = Y_train_fold[i]
    y_test = Y_test_fold[i]

    split_idx = int(len(X_train) * 0.8)

    X_tr = X_train[:split_idx]
    X_val = X_train[split_idx:]
    
    y_tr = y_train[:split_idx]
    y_val = y_train[split_idx:]

    print("X_train.shape", X_tr.shape)
    print("X_val.shape", X_val.shape)
    print("X_test.shape", X_test.shape)

    model = create_fedformer_model(X_train.shape[1:], y_train.shape[1])

    model.fit(
        X_tr, y_tr,
        epochs=epoch_global,
        batch_size=batch_size_global,
        validation_data=(X_val, y_val),
        callbacks=[early_stopping, Regularisasi],
        verbose=1
    )

    y_pred_prob = model.predict(X_test).ravel()
    y_pred = (y_pred_prob >= threshold).astype(int)

    y_true[i] = y_test
    y_preds[i] = y_pred
    y_probs[i] = y_pred_prob


# =========================
# EVALUATION
# =========================
df_cv_fedformer, df_meancv_fedformer, y_true_fedformer, y_prob_fedformer = eval_fold(y_true, y_preds, y_probs)
```

## 7) VARIATIONAL AUTOENCODER

```python
print('='*10, 'FORWARD CHAINING CROSS VALIDATION -VAE- ','='*10)

y_true = {}
y_preds = {}
y_probs = {}

import tensorflow as tf
from tensorflow.keras.layers import (
    Input, Dense, Dropout, Flatten
)
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam


# =========================
# SAMPLING LAYER
# =========================
class Sampling(tf.keras.layers.Layer):
    def call(self, inputs):
        z_mean, z_log_var = inputs

        z_log_var = tf.clip_by_value(z_log_var, -10.0, 10.0)

        epsilon = tf.random.normal(shape=tf.shape(z_mean))
        return z_mean + tf.exp(0.5 * z_log_var) * epsilon


# =========================
# KL LOSS LAYER (FIX ERROR)
# =========================
class KLDivergenceLayer(tf.keras.layers.Layer):
    def call(self, inputs):
        z_mean, z_log_var = inputs

        # 🔥 CLIP biar gak explode
        z_log_var = tf.clip_by_value(z_log_var, -10.0, 10.0)

        kl_loss = -0.5 * tf.reduce_mean(
            1 + z_log_var - tf.square(z_mean) - tf.exp(z_log_var)
        )

        self.add_loss(0.001 * kl_loss)  # kecilin weight

        return z_mean, z_log_var


# =========================
# MODEL
# =========================
def create_vae_model(input_shape, output_dim):
    inputs = Input(shape=input_shape)

    # =========================
    # FLATTEN
    # =========================
    x = Flatten()(inputs)

    # =========================
    # ENCODER
    # =========================
    x = Dense(64, activation='relu')(x)
    x = Dropout(0.1)(x)

    z_mean = Dense(16)(x)
    z_log_var = Dense(16)(x)

    # =========================
    # KL LOSS INJECTION
    # =========================
    z_mean, z_log_var = KLDivergenceLayer()([z_mean, z_log_var])

    # =========================
    # SAMPLING
    # =========================
    z = Sampling()([z_mean, z_log_var])

    # =========================
    # CLASSIFIER HEAD
    # =========================
    x = Dense(32, activation='relu')(z)
    x = Dropout(0.1)(x)
    outputs = Dense(output_dim, activation='sigmoid')(x)

    model = Model(inputs, outputs)

    model.compile(
        optimizer=Adam(learning_rate=0.001),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    return model


# =========================
# TRAINING LOOP (SAMA)
# =========================
for i in range(5):
    print()
    print(f'================= FOLD = {i} ==============')
    print()

    X_train = X_train_fold[i]
    X_test = X_test_fold[i]
    y_train = Y_train_fold[i]
    y_test = Y_test_fold[i]

    split_idx = int(len(X_train) * 0.8)

    X_tr = X_train[:split_idx]
    X_val = X_train[split_idx:]
    
    y_tr = y_train[:split_idx]
    y_val = y_train[split_idx:]

    print("X_train.shape", X_tr.shape)
    print("X_val.shape", X_val.shape)
    print("X_test.shape", X_test.shape)

    model = create_vae_model(X_train.shape[1:], y_train.shape[1])

    model.fit(
        X_tr, y_tr,
        epochs=epoch_global,
        batch_size=batch_size_global,
        validation_data=(X_val, y_val),
        callbacks=[early_stopping, Regularisasi],
        verbose=1
    )

    # =========================
    # PREDICTION
    # =========================
    y_pred_prob = model.predict(X_test).ravel()
    y_pred = (y_pred_prob >= threshold).astype(int)

    # =========================
    # SAVE
    # =========================
    y_true[i] = y_test
    y_preds[i] = y_pred
    y_probs[i] = y_pred_prob


# =========================
# EVALUATION
# =========================
df_cv_vae, df_meancv_vae, y_true_vae, y_prob_vae = eval_fold(y_true, y_preds, y_probs)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import VotingClassifier
from sklearn.multioutput import MultiOutputClassifier
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Dropout, Concatenate
from tensorflow.keras.optimizers import Adam
from sklearn.metrics import accuracy_score
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Simulasi data dummy
# n_step = 12
n_fitur_asli = n_fitur_fix
feature_names_asli = ['Production', 'Importation', 'Demand', 'Price', 'Jumlah Penduduk',
       'HBKN', 'Latitude', 'Longitude', 'Economy Growth (%)']
seed = 42

# Index fitur "Production"
# idx_production = feature_names_asli.index("Production", "Supply" , "Longitude", "Latitude" )
top_features = ["Production"]#, "Jumlah Penduduk", "Demand"]
idx_top_features = [feature_names_asli.index(f) for f in top_features]

print('='*10, 'FORWARD CHAINING CROSS VALIDATION','='*10)

y_true = {}
y_preds = {}
y_probs = {}

# Fungsi untuk membuat conditional autoencoder
def create_conditional_autoencoder_encoder(input_dim, cond_dim, dropout_rate=0.1):
    input_x = Input(shape=(input_dim,))
    input_cond = Input(shape=(cond_dim,))
    merged_input = Concatenate()([input_x, input_cond])

    # Encoder
    x = Dense(256, activation='relu')(merged_input)
    x = Dropout(dropout_rate)(x)
    x = Dense(128, activation='relu')(x)
    x = Dropout(dropout_rate)(x)
    x = Dense(64, activation='relu')(x)
    x = Dropout(dropout_rate)(x)
    x = Dense(32, activation='relu')(x)
    x = Dropout(dropout_rate)(x)
    x = Dense(20, activation='relu')(x)
    x = Dropout(dropout_rate)(x)
    encoded = Dense(16, activation='relu')(x)

    # Decoder
    x = Concatenate()([encoded, input_cond])  # tetap conditional
    x = Dropout(dropout_rate)(x)
    x = Dense(20, activation='relu')(x)
    x = Dropout(dropout_rate)(x)
    x = Dense(32, activation='relu')(x)
    x = Dropout(dropout_rate)(x)
    x = Dense(64, activation='relu')(x)
    x = Dropout(dropout_rate)(x)
    x = Dense(128, activation='relu')(x)
    x = Dropout(dropout_rate)(x)
    x = Dense(256, activation='relu')(x)
    decoded = Dense(input_dim, activation='sigmoid')(x)  # atau 'linear' jika data tidak [0,1]

    autoencoder = Model(inputs=[input_x, input_cond], outputs=decoded)
    encoder = Model(inputs=[input_x, input_cond], outputs=encoded)

    return autoencoder, encoder



# Melakukan looping untuk 5-fold
for i in range(5):
    print(f"\n===== FOLD {i+1} =====")

    # Ambil data
    X_train = X_train_fold[i]
    X_test = X_test_fold[i]
    y_train = Y_train_fold[i]
    y_test = Y_test_fold[i]

    # Flatten fitur (samples, n_step * n_features)
    X_train_flat = X_train.reshape(X_train.shape[0], -1)
    X_test_flat = X_test.reshape(X_test.shape[0], -1)

    # Ambil y_cond dari fitur "Production" per time step
    # y_cond_train = X_train[:, :, idx_production]  # shape: (samples, n_step)
    # y_cond_test = X_test[:, :, idx_production]

    y_cond_train = X_train[:, :, idx_top_features]  # (samples, n_step, 4)
    y_cond_test = X_test[:, :, idx_top_features]

    y_cond_train = y_cond_train.reshape(y_cond_train.shape[0], -1)
    y_cond_test = y_cond_test.reshape(y_cond_test.shape[0], -1)

    # ==== Conditional Autoencoder ====
    input_dim = X_train_flat.shape[1]
    cond_dim = y_cond_train.shape[1]

    # Panggil fungsi untuk membuat autoencoder dan encoder
    autoencoder, encoder_model = create_conditional_autoencoder_encoder(input_dim, cond_dim)
 
    # Compile dan fit model autoencoder
    autoencoder.compile(optimizer=Adam(0.001), loss='mse')
    autoencoder.fit(
        [X_train_flat, y_cond_train],
        X_train_flat,
        validation_split=0.2,
        shuffle=False,
        epochs=150,
        batch_size=32,
        verbose=1,
        callbacks=[early_stopping, Regularisasi]
    )
    
    # ================= LATENT =================
    X_train_latent = encoder_model.predict([X_train_flat, y_cond_train])
    X_test_latent = encoder_model.predict([X_test_flat, y_cond_test])
    
    print(f"Latent shape (before): {X_train_latent.shape}")
    
    # ================= SCALE (WAJIB) =================
    scaler = StandardScaler()
    X_train_latent_scaled = scaler.fit_transform(X_train_latent)
    X_test_latent_scaled = scaler.transform(X_test_latent)
    
    # ================= PCA FILTER =================
    pca = PCA(n_components=3, random_state=seed)
    
    X_train_pca = pca.fit_transform(X_train_latent_scaled)
    X_test_pca = pca.transform(X_test_latent_scaled)

    print("Explained variance ratio:", pca.explained_variance_ratio_)
    print("Total variance kept:", np.sum(pca.explained_variance_ratio_))
    
    print(f"Latent shape (after PCA): {X_train_pca.shape}")
    
    # ================= CONCAT (PAKAI PCA, BUKAN LATENT ASLI) =================
    encoded_df = pd.DataFrame(X_train_flat, columns=[f"enc_{i}" for i in range(X_train_flat.shape[1])])
    latent_df = pd.DataFrame(X_train_pca, columns=[f"pca_{i}" for i in range(X_train_pca.shape[1])])
    X_train_jadi = pd.concat([encoded_df, latent_df], axis=1)
    
    encoded_test_df = pd.DataFrame(X_test_flat, columns=[f"enc_{i}" for i in range(X_test_flat.shape[1])])
    latent_test_df = pd.DataFrame(X_test_pca, columns=[f"pca_{i}" for i in range(X_test_pca.shape[1])])
    X_test_jadi = pd.concat([encoded_test_df, latent_test_df], axis=1)
    
    print(f"latent_stacking : {X_train_jadi.shape}")
    
    # ================= MODEL =================
    xgb = XGBClassifier(
        n_estimators=50,
        max_depth=5,
        use_label_encoder=False,
        eval_metric='logloss',
        random_state=seed
    )
    
    catboost = CatBoostClassifier(
        iterations=50,
        learning_rate=0.1,
        depth=5,
        verbose=0,
        random_seed=seed
    )
    
    voting_clf = VotingClassifier(
        estimators=[('xgb', xgb), ('catboost', catboost)],
        voting='soft',
        n_jobs=-1
    )
    
    hybrid_model = MultiOutputClassifier(voting_clf)
    hybrid_model.fit(X_train_jadi, y_train)
    
    # ================= PROBA =================
    # voting_models = hybrid_model.estimators_
    y_prob_list = []
    
    for model in hybrid_model.estimators_:
        prob = model.predict_proba(X_test_jadi)[:, 1]  # 🔥 langsung dari soft voting
        y_prob_list.append(prob)
    
    # shape: (n_samples, n_outputs)
    y_prob = np.array(y_prob_list).T
    
    
    # ================= APPLY THRESHOLD =================
    threshold = threshold  # 🔥 bisa kamu ubah (0.5 / 0.7 dll)
    
    y_pred = (y_prob >= threshold).astype(int)
    
    y_true[i] = y_test
    y_preds[i] = y_pred
    y_probs[i] = y_prob
    
    # ================= EVAL =================
df_cv_ae_cae3, df_meancv_ae_cae3, y_true_ae_cae3, y_prob_ae_cae3 = eval_fold(y_true, y_preds, y_probs)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import precision_recall_curve, auc

# =========================
# HELPER FUNCTION
# =========================
def compute_pr(y_true, y_prob):
    precision, recall, _ = precision_recall_curve(y_true, y_prob)
    pr_auc = auc(recall, precision)
    return precision, recall, pr_auc

# =========================
# GROUPING MODELS
# =========================
dl_models = {
    "CNN": (y_true_cnn, y_prob_cnn),
    "LSTM": (y_true_lstm, y_prob_lstm),
    "GRU": (y_true_gru, y_prob_gru),
    "TCN": (y_true_tcn, y_prob_tcn),
    "Informer": (y_true_informer, y_prob_informer),
    "FedFormer": (y_true_fedformer, y_prob_fedformer),
    "VAE": (y_true_vae, y_prob_vae),
}

tree_models = {
    "XGB": (y_true_xgb, y_prob_xgb),
    "CAT": (y_true_cat, y_prob_cat),
    "XGB-CAT": (y_true_ae_search, y_prob_ae_search),
}

ae_models = {
    "CAE2 + XGB-CAT": (y_true_ae_cae2, y_prob_ae_cae2),
    "CAE3 + XGB-CAT": (y_true_ae_cae3, y_prob_ae_cae3),
    "AE Basic1": (y_true_ae_basic1, y_prob_ae_basic1),
    "AE Basic2": (y_true_ae_basic2, y_prob_ae_basic2),
    "AE Basic3": (y_true_ae_basic3, y_prob_ae_basic3),
    "Seq2Seq": (y_true_ae_seq2seq, y_prob_ae_seq2seq),
}

# =========================
# FUNCTION PLOT GROUP
# =========================
def plot_group(models_dict, title, filename):

    plt.figure(figsize=(8,6))

    for name, (y_true, y_prob) in models_dict.items():
        precision, recall, pr_auc = compute_pr(y_true, y_prob)
        plt.plot(recall, precision, label=f"{name} (AUC={pr_auc:.3f})")

    plt.xlabel("Recall")
    plt.ylabel("Precision")
    plt.title(title)
    plt.legend(fontsize=8)
    plt.grid(True)

    plt.tight_layout()

    # 🔥 SAVE DI SINI
    plt.savefig(filename, dpi=800, bbox_inches='tight')

    plt.show()


# =========================
# FUNCTION AVERAGE CURVE
# =========================
def plot_average(models_dict):

    all_precisions = []

    for name, (y_true, y_prob) in models_dict.items():
        precision, recall, _ = compute_pr(y_true, y_prob)

        recall_interp = np.linspace(0,1,100)
        precision_interp = np.interp(recall_interp, recall[::-1], precision[::-1])

        all_precisions.append(precision_interp)

    mean_precision = np.mean(all_precisions, axis=0)
    mean_recall = np.linspace(0,1,100)

    return mean_recall, mean_precision


# =========================
# === FIGURE 1 ===
# =========================
plot_group(dl_models, "PR Curve - Baseline Deep Learning", "PR-CURVE-DL-BASELINE.jpg")

# =========================
# === FIGURE 2 ===
# =========================
plot_group(tree_models, "PR Curve - Tree Based Models", "PR-CURVE-TREE-BASED.jpg")

# =========================
# === FIGURE 3 ===
# =========================
plot_group(ae_models, "PR Curve - AE Family Models", "PR-CURVE-AE's.jpg")

# =========================
# === FIGURE 4 (COMPARISON) ===
# =========================
plt.figure(figsize=(8,6))

r_dl, p_dl = plot_average(dl_models)
r_tree, p_tree = plot_average(tree_models)
r_ae, p_ae = plot_average(ae_models)

plt.plot(r_dl, p_dl, label="Baseline DL", linewidth=2)
plt.plot(r_tree, p_tree, label="Tree-Based", linewidth=2)
plt.plot(r_ae, p_ae, label="AE Family (Proposed)", linewidth=2)

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("PR Curve Comparison (Averaged)")
plt.legend()
plt.grid(True)

plt.tight_layout()

# 🔥 SAVE
plt.savefig("fig4_PR-CURVE-AVG.jpg", dpi=800, bbox_inches='tight')

plt.show()
```
